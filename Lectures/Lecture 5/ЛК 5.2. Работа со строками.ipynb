# Работа со строками

## Литералы строк
Работа со строками в ```Python``` очень удобна. Существует несколько литералов строк, которые мы сейчас и рассмотрим.

### Строки в апострофах и в кавычках

```python
s = 'spam"s'
s = "spam's"
```

Строки в апострофах и в кавычках - одно и то же. Причина наличия двух вариантов в том, чтобы позволить вставлять в литералы строк символы кавычек или апострофов, не используя экранирование.

### "Сырые" строки - подавляют экранирование

Если перед открывающей кавычкой стоит символ '```r```' (в любом регистре), то механизм экранирования отключается.

```python
s = r'C:\newt.txt'
```

Но, несмотря на назначение, "сырая" строка не может заканчиваться символом обратного слэша. Пути решения:

```python
s = r'\n\n\\'[:-1]
s = r'\n\n' + '\\'
s = '\\n\\n'
```

````python
s = "C:\newt.txt"

print(s)
````

````python
s = "C:\\newt.txt"

print(s)
````

````python
s = r"C:\newt.txt"

print(s)
````

````python
s = r"\n\n\"

print(s)
````

````python
s = r"\n\n"

print(s)
````

### Строки в тройных апострофах или кавычках

Главное достоинство строк в тройных кавычках в том, что их можно использовать для записи многострочных блоков текста. Внутри такой строки возможно присутствие кавычек и апострофов, главное, чтобы не было трех кавычек подряд.

````python
c = '''это очень большая
строка, многострочный
блок текста'''

print(c)
````

````python
c = """это очень большая
строка, многострочный
блок текста"""

print(c)
````

````python
c = "это очень большая\nстрока, многострочный\nблок текста"

print(c)
````

# Базовые операции

* **Конкатенация (сложение)**

```python
>>> s1 = 'spam'
>>> s2 = 'eggs'
>>> print(s1 + s2)
'spameggs'
```

* **Дублирование строки**

```python
>>> print('spam' * 3)
spamspamspam
```

* **Длина строки (функция len)**

```python
>>> len('spam')
4
```

* **Доступ по индексу**

```python
>>> s = 'spam'
>>> s[0]
's'
>>> s[2]
'a'
>>> s[-2]
'a'
```

Как видно из примера, в ```Python``` возможен и доступ по отрицательному индексу, при этом отсчет идет от конца строки.

* **Извлечение среза**

Оператор извлечения среза: ```[X:Y]```. ```X``` – начало среза, а ```Y``` – окончание;

символ с номером ```Y``` в срез не входит. По умолчанию первый индекс равен ```0```, а второй - ```длине строки```.

```python
>>> s = 'spameggs'
>>> s[3:5]
'me'
>>> s[2:-2]
'ameg'
>>> s[:6]
'spameg'
>>> s[1:]
'pameggs'
>>> s[:]
'spameggs'
```

Кроме того, можно задать шаг, с которым нужно извлекать срез.

``` python
>>> s[::-1]
'sggemaps'
>>> s[3:5:-1]
''
>>> s[2::2]
'aeg'
```

## Другие функции и методы строк

При вызове методов необходимо помнить, что строки в ```Python``` относятся к категории неизменяемых последовательностей, то есть все функции и методы могут лишь создавать новую строку.

```python
>>> s = 'spam'
>>> s[1] = 'b'
Traceback (most recent call last):
  File "", line 1, in
    s[1] = 'b'
TypeError: 'str' object does not support item assignment
>>> s = s[0] + 'b' + s[2:]
>>> s
'sbam'
```

Поэтому все строковые методы возвращают новую строку, которую потом следует присвоить переменной.

### Наличие подстроки и поиск позиции

| Задача | Инструмент | Результат |
| :--- | :--- | :--- |
| Есть ли подстрока? | `part in s` | `True` / `False` |
| С чего начинается строка? | `s.startswith(prefix)` | `True` / `False` |
| Чем заканчивается строка? | `s.endswith(suffix)` | `True` / `False` |
| Где первое вхождение? | `s.find(part)` | Индекс или `-1` |
| Где последнее вхождение? | `s.rfind(part)` | Индекс или `-1` |
| Найти обязательную подстроку | `s.index(part)` | Индекс или `ValueError` |
| Сколько вхождений? | `s.count(part)` | Число непересекающихся вхождений |

> Ловушка: не используйте `if s.find(part)` для проверки наличия. Индекс `0` трактуется как ложь, а `-1` — как истина. Для проверки наличия используйте `in`.

````python
filename = "survey_2026_points.csv"
print("Есть points:", "points" in filename)
print("Префикс:", filename.startswith("survey_"))
print("Расширение:", filename.endswith(".csv"))
print("Первое подчёркивание:", filename.find("_"))
print("Последнее подчёркивание:", filename.rfind("_"))
print("Нет подстроки:", filename.find("height"))
print("Число разделителей:", filename.count("_"))
````

### Очистка, замена и регистр

`strip()` удаляет пробельные символы по краям, `replace()` заменяет подстроки, `lower()` и `upper()` меняют регистр. Для сравнения текста без учёта регистра предусмотрен `casefold()`.

Аргумент `strip(chars)` задаёт набор удаляемых символов, а не цельный префикс или суффикс. Для удаления точного префикса и суффикса используйте `removeprefix()` и `removesuffix()` (Python 3.9+).

````python
raw_name = "  Пункт А-01\n"
print(repr(raw_name.strip()))
print(raw_name.strip().lower())
print("12,345".replace(",", "."))
print("а-а-а".replace("а", "б", 1))
print("ПУНКТ".casefold() == "пункт".casefold())

print("Набор символов:", "ab_data_ba".strip("ab_"))
print("Точный префикс:", "ab_data_ba".removeprefix("ab_"))
print("Точный суффикс:", "points.csv".removesuffix(".csv"))
````

### Проверка содержимого

Методы `isalpha()`, `isdigit()`, `isalnum()` и `isspace()` проверяют содержимое строки. Пустая строка для этих четырёх методов даёт `False`.

`isdigit()` — не универсальная проверка числа: знак, десятичная точка и запятая не являются цифрами. Некоторые Unicode-символы проходят `isdigit()`, но не подходят для `int()`. Для реального преобразования используйте `int()` / `float()` с обработкой `ValueError`.

````python
samples = ["123", "-12", "12.5", "12,5", "²", "", "А01"]
for value in samples:
    print(f"{value!r:>8} → isdigit(): {value.isdigit()}")

value = "-12,5"
try:
    number = float(value.replace(",", "."))
    print("Число:", number)
except ValueError:
    print("Не удалось преобразовать:", repr(value))
````

## Разбиение и сборка

`split()` превращает строку в список частей. `separator.join(parts)` выполняет обратную задачу: соединяет строки с выбранным разделителем.

- `split()` без аргумента объединяет последовательности пробельных символов и не создаёт пустых частей по краям.
- `split(" ")` использует ровно один пробел как разделитель, поэтому соседние пробелы создают пустые части.
- `splitlines()` разбивает многострочный текст на строки.
- `partition(sep)` делит по первому разделителю и возвращает три части: до разделителя, разделитель, после него.

````python
text = "  А-01   12.345  "
print("По пробельным символам:", text.split())
print("По одному пробелу:     ", text.split(" "))

record = "А-01;100.250;200.500;12.345"
parts = record.split(";")
print("Поля:", parts)
print("Новый разделитель:", " | ".join(parts))
print("Первая граница:", "height=12.345".partition("="))
print("Строки:", "А-01\nА-02\n".splitlines())
````

> Граница применимости: `split(";")` подходит для простых записей с заранее известным устройством. Если разделитель может находиться внутри поля в кавычках, для CSV используйте модуль `csv`, а не ручное разбиение.

Самопроверка: почему `";".join(["А-01", "12.345"])` допустимо, а список `["А-01", 12.345]` сначала требует преобразования числа в строку?

# Форматирование строк

###  F-строки

Префикс `f` позволяет вставлять значения и выражения в фигурных скобках. После двоеточия указывается формат: например, `.3f` выводит число с тремя знаками после точки.

Форматирование задаёт текстовое представление числа, но не изменяет исходное числовое значение.

````python
station = "А-01"
height = 12.34567
reference_height = 12.34000

print(f"Пункт {station}: H = {height:.3f} м")
print(f"Отклонение: {(height - reference_height) * 1000:+.1f} мм")
print(f"Исходное число: {height}")
print(f"Фигурные скобки: {{H}} = {height:.3f}")
````

### Ширина поля и выравнивание

| Формат | Назначение |
| :--- | :--- |
| `.3f` | Три знака после десятичной точки |
| `+.2f` | Явный знак и два знака после точки |
| `>10` | Выравнивание вправо в поле шириной не менее 10 |
| `<10` | Выравнивание влево |
| `^10` | Выравнивание по центру |
| `03d` | Целое число, не менее трёх позиций, заполнение нулями |

Ширина поля — минимальная, а не максимальная: длинное значение не обрезается.

````python
stations = [("А-01", 12.34567), ("А-02", -9.8), ("Б-103", 105.1264)]
print(f"{'Пункт':<10} {'Высота, м':>12}")
print("─" * 23)

for name, height in stations:
    print(f"{name:<10} {height:>+12.3f}")
print(f"Имя файла: station_{7:03d}.txt")
````

### Составной текст и другие способы форматирования

В скобках соседние строковые литералы объединяются. Префикс `f` нужен у каждого литерала, в котором требуется подстановка. Для настоящего перевода строки используйте `\n` или тройные кавычки.

`str.format()` удобен, когда шаблон хранится отдельно от значений. Форматирование через `%` полезно узнавать при чтении существующего кода; здесь основной способ — f-строки.

````python
name = "А-01"
height = 12.34567
message = (
    f"Пункт: {name}. "
    f"Высота: {height:.3f} м. "
    "Обработка завершена."
)
print(message)

template = "Пункт {name}: {height:.3f} м"
print(template.format(name=name, height=height))
print("Пункт %s: %.3f м" % (name, height))
````

## Сквозной пример

Учебный формат записи: `пункт;X;Y;H`. Координаты и высота записаны в метрах, десятичный разделитель — запятая или точка. Поля не содержат кавычек и внутренних точек с запятой.

Порядок работы:
1. Удалить пробельные символы по краям записи.
2. Разделить поля и проверить их количество.
3. Очистить имя пункта и преобразовать числа.
4. Сформировать строку отчёта с единым числом десятичных знаков.

Здесь проверяются структура записи и возможность числового преобразования, но не геодезическая корректность координат.

````python
raw_record = "  а-01 ; 100,250 ; 200,500 ; 12,3456  "
fields = raw_record.strip().split(";")

if len(fields) != 4:
    raise ValueError("Ожидаются четыре поля: пункт;X;Y;H")

station = fields[0].strip().upper()
if not station:
    raise ValueError("Имя пункта не должно быть пустым")

x = float(fields[1].strip().replace(",", "."))
y = float(fields[2].strip().replace(",", "."))
height = float(fields[3].strip().replace(",", "."))

result = f"{station}: X = {x:.3f} м; Y = {y:.3f} м; H = {height:.3f} м"
print(result)
````

## Метод ```format```

Иногда (а точнее, довольно часто) возникают ситуации, когда нужно сделать строку, подставив в неё некоторые данные, полученные в процессе выполнения программы (пользовательский ввод, данные из файлов и т. д.). Подстановку данных можно сделать с помощью форматирования строк. Форматирование можно сделать с помощью оператора ```%```, либо с помощью метода ```format```.

Если для подстановки требуется только один аргумент, то значение - сам аргумент:

```python
>>> 'Hello, {}!'.format('Vasya')
'Hello, Vasya!'
```

А если несколько, то значениями будут являться все аргументы со строками подстановки (обычных или именованных):

```python
>>> '{0}, {1}, {2}'.format('a', 'b', 'c')
'a, b, c'
>>> '{}, {}, {}'.format('a', 'b', 'c')
'a, b, c'
>>> '{2}, {1}, {0}'.format('a', 'b', 'c')
'c, b, a'
>>> '{2}, {1}, {0}'.format(*'abc')
'c, b, a'
>>> '{0}{1}{0}'.format('abra', 'cad')
'abracadabra'
>>> 'Coordinates: {latitude}, {longitude}'.format(latitude='37.24N', longitude='-115.81W')
'Coordinates: 37.24N, -115.81W'
>>> coord = {'latitude': '37.24N', 'longitude': '-115.81W'}
>>> 'Coordinates: {latitude}, {longitude}'.format(**coord)
'Coordinates: 37.24N, -115.81W'
```

### Unicode и байты: короткое дополнение

`ord()` возвращает кодовую точку Unicode для одного символа, а `chr()` выполняет обратное преобразование. Это не только ASCII.

`str` хранит текст, `bytes` — байты. Кодирование (`encode`) превращает текст в байты, декодирование (`decode`) — байты в текст. `len(str)` считает кодовые точки, не байты и не обязательно видимые пользователю символы.

````python
symbol = "А"
print("Кодовая точка:", ord(symbol))
print("Обратно:", chr(ord(symbol)))

text = "Пункт"
data = text.encode("utf-8")
print("Текст:", text)
print("Байты:", data)
print("Кодовых точек:", len(text), "| Байтов UTF-8:", len(data))
````
