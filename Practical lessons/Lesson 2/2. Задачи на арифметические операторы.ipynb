# Задачи на арифметические операторы

### Обозначения операторов


| **Оператор** |      **Значение**     | **Пример**  |
|--------------|:---------------------:|-------------|
|       +      |        Сложение       | 2 + 2 = 4   |
|       -      |       Вычитание       | 4 - 2 = 2   |
|       *      |       Умножение       | 2 * 2 = 4   |
|       /      |        Деление        | 4 / 2 = 2.0 |
|      **      |  Возведение в степень | 2 ** 3 = 8  |
|       %      |   Остаток от деления  | 5 % 2 = 1   |
|      //      | Целочисленное деление | 5 // 2 = 2  |

### Приоретет операторов

**Оператор**|**Название**
:-----:|:-----:
( )|Скобки
**|Возведение в степень
+x, -x, ~x|Унарные плюс, минус и битовое отрицание
*, /, //, %|Умножение, деления, взятие остатка
+, -|Сложение и вычитание
<<, >>|Битовые сдвиги
&|Битовое И
^|Битовое исключающее ИЛИ (XOR)
\||Битовое ИЛИ
==, !=, >, >=, <, <=, is, is not, in, not in|Сравнение, проверка идентичности, проверка вхождения
not|Логическое НЕ
and|Логическое И
or|Логическое ИЛИ
=|Операторы присваивания

## Задача 2.3
Автомобиль проехал первый участок длиной `s1` километров со скоростью `v1` км/ч, а второй участок длиной `s2` километров со скоростью `v2` км/ч.

Напишите программу, которая определяет среднюю скорость автомобиля за весь путь.
 
 Используйте формулу:
 
$$
 v_{\text{ср}} =
 \frac{s_1 + s_2}{
 (s_1 / v_1) + (s_2 / v_2)
 }.
$$
